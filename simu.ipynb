```python
import simpy 


def customer(env, name, barista):
    print(f"⏱ {env.now:2} min: {name} arrives")

    with barista.request() as request:
        yield request

        print(f"☕ {env.now:2} min: {name} starts being served")
        yield env.timeout(3)
        print(f"✅ {env.now:2} min: {name} gets their coffee")

def arrivals(env, barista):
    for name in ["Alice", "Bob", "Charlie"]:
        env.process(customer(env, name, barista))
        yield env.timeout(1)

env = simpy.Environment()
barista = simpy.Resource(env, capacity=1)

env.process(arrivals(env, barista))
env.run()
```
